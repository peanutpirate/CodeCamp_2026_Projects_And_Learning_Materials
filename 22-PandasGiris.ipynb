## Series

```python
# Pandasa veri biliminin exceli derler. verileri manipule edebilmek için hem de verileri akla mantığa yatkın gösterebilmek için kullanıyoruz. 
```

```python
# serileri sözlüklerden listelerden numpy dizilerinden oluşturabiliriz.
```

```python
benimSozlugum = {"Atil" : 50, "Zeynep" : 40, "Mehmet" : 30}
```

```python
pd.Series(benimSozlugum)
```

```python
benimYaslarim = [50,40,30]
benimIsimlerim = ["Atil","Zeynep","Mehmet"]
```

```python
pd.Series(benimYaslarim)
```

```python
pd.Series(benimYaslarim, benimIsimlerim)
```

```python
pd.Series(data=benimYaslarim, index=benimIsimlerim)
```

```python
numpyDizisi = np.array([50, 40, 30])
```

```python
pd.Series(numpyDizisi)
```

```python
pd.Series(["Atil","Atlas","Osman"], [1,2,3])
```

```python
yarismaSonucu1 = pd.Series([10,5,1],["Atil","Atlas","Osman"])
```

```python
yarismaSonucu1
```

```python
yarismaSonucu2
```
