# 🧩 KULLANILAN PYTHON KONULARI
Bu projede yalnızca aşağıdaki Python konuları kullanılmaktadır:
- Fonksiyonlar
- Girdi / Çıktı işlemleri (input, print)
- Değişken sayıda argüman (*args)
- Önemli gömülü fonksiyonlar:
  - sum()
  - len()
  - round()
  - max()
  - min()
- Kapsam (local / global)
- Hata yakalama (try / except)
- Hesap makinesi mantığı ve akış kontrolü

```python
# 2. Harcama Alma Fonksiyonu
def get_expenses():
    # Kullanıcıdan güvenli şekilde harcama listesi alır
    expenses = []

    # Harcama sayısını al
    while True:
        try:
            n = int(input("Kaç harcama gireceksiniz? "))
            if n <= 0:
                print("Lütfen 1 veya daha büyük bir sayı girin.")
            else:
                break
        except ValueError:
            print("Geçersiz giriş. Lütfen tam sayı girin.")

    i = 0
    # Harcamaları tek döngüyle al
    while i < n:
        try:
            val = float(input(f"{i+1}. harcama tutarı: "))
            if val < 0:
                print("Harcama tutarı negatif olamaz.")
            else:
                expenses.append(val)
                i += 1
        except ValueError:
            print("Geçersiz giriş. Lütfen sayısal bir değer girin.")

    return expenses

harcamalar = get_expenses()
print("Girilen harcamalar:", harcamalar)
```

```python
# 3. Toplam ve Ortalama Fonksiyou
def calculate_summary(expenses):
    # Toplam ve ortalama harcamayı hesaplar

    if len(expenses) == 0:
        print("Harcama listesi boş.")
        return

    total = sum(expenses)
    average = total / len(expenses)

    print("Toplam Harcama:", round(total, 2))
    print("Ortalama Harcama:", round(average, 2))

harcamalar = get_expenses()
calculate_summary(harcamalar)
```

```python
# 4.*Args ile Esnek Hesaplama
def flexible_total(*args):
    # Girilen tüm argümanları toplar ve argüman sayısını gösterir

    if len(args) == 0:
        print("Hiç harcama girilmedi.")
        return

    total = sum(args)

    print("Girilen harcama sayısı:", len(args))
    print("Toplam Harcama:", round(total, 2))
  # Örnek
flexible_total(120, 300, 75.5)
```

```python
# 5. En Büyük/En Küçük Harcama
def find_extremes(expenses):
    # En büyük ve en küçük harcamayı bulur

    if len(expenses) == 0:
        print("Harcama listesi boş.")
        return

    max_expense = max(expenses)
    min_expense = min(expenses)

    print("En Büyük Harcama:", round(max_expense, 2))
    print("En Küçük Harcama:", round(min_expense, 2))

harcamalar = get_expenses()
find_extremes(harcamalar)
```

```python
# 6. Kapsam (Scope) Gösterimi - Önemli!

# currency değişkeni global olarak tanımlanmıştır
# Fonksiyon içinde bu değişken okunabilir

# Ancak fonksiyon içinde currency = "USD" gibi bir atama yapılırsa
# Python bunu global değil, local değişken olarak algılar

# Bu nedenle:
# - Global değişken okunabilir
# - Global değişken doğrudan değiştirilemez
# - Aynı isimle yapılan atama local scope oluşturur

# Amaç:
# Local vs Global scope farkını göstermek

currency = "TL"

def show_currency():
    # Global değişkeni OKUMA
    print("Fonksiyon içinden okunan para birimi:", currency)

    # currency = "USD"
    # print("Fonksiyon içinde değiştirilen para birimi:", currency)

show_currency()

print("Fonksiyon çağrısından sonra global currency:", currency)
```

```python
# 7. Ana Akış (Main Flow)
print("=== Smart Budget Assistant (v4) ===")

# 1️⃣ Harcamaları al
harcamalar = get_expenses()

# 2️⃣ Toplam ve ortalama
print("\n--- Harcama Özeti ---")
calculate_summary(harcamalar)

# 3️⃣ En büyük / en küçük harcama
print("\n--- Harcama Limitleri ---")
find_extremes(harcamalar)

# 4️⃣ Esnek toplam (args gösterimi)
print("\n--- Esnek Toplam (Args Demo) ---")
flexible_total(*harcamalar)

print("\nProgram tamamlandı ✅")
```
