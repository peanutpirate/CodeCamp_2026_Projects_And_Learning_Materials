```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
yasListesi = [10,20,30,30,30,40,50,60,70,75]
```

```python
kiloListesi = [20,60,80,85,86,87,70,90,95,90]
```

```python
numpyYasListesi = np.array(yasListesi)
numpyKiloListesi = np.array(kiloListesi)
```

```python
numpyKiloListesi
```

```python
numpyYasListesi
```

```python
plt.plot(numpyYasListesi,numpyKiloListesi,"g")
plt.xlabel("Yas")
plt.ylabel("Kilo")
plt.title("Kilonu yaşa Göre Değişim Grafiği")
# plt.show()
```

```python
numpyDizisi1 = np.linspace(0,10,20) # 0 ile 10 arasında 20 tane veri oluştur. 
```

```python
numpyDizisi1
```

```python
numpyDizisi2 = numpyDizisi1 ** 3 
```

```python
numpyDizisi2
```

```python
plt.plot(numpyDizisi1,numpyDizisi2,"g*-")
# çizgileri kesik kesik --
# yıldız tire li noktalı- Veri noktaları önemliyse
```

```python
plt.subplot(1,2,1)
# iki ayrı grafik sıralı ya da alt alta olabilir.
plt.plot(numpyDizisi1,numpyDizisi2,"r*-")
plt.subplot(1,2,2) # 1 sıra olucak 2 kolon olucak şu an 2. grafiği çiziyorum
plt.plot(numpyDizisi2,numpyDizisi1,"g--")
```

```python
# plt import ettiğimiz kütüphane
# o sınıftan obje oluşturup oluşturduğumuz obje üzerinden işlem yapıcaz. 
benimFigure = plt.figure()
figureAxes = benimFigure.add_axes([0.4,0.4,0.6,0.6]) # width height genişlik yükselik
figureAxes.plot(numpyDizisi1,numpyDizisi2,"g")
figureAxes.set_xlabel("X Ekseni Veri İsmi")
figureAxes.set_ylabel("Y Ekseni Veri İsmi")
figureAxes.set_title("Grafik Başlığı")
```

```python
figure2 = plt.figure()

eksen1 = figure2.add_axes([0.1,0.1,0.7,0.7])
eksen2 = figure2.add_axes([0.2,0.3,0.3,0.3])

eksen1.plot(numpyDizisi1,numpyDizisi2,"g")
eksen1.set_xlabel("X Ekseni")
eksen1.set_ylabel("Y Ekseni")
eksen1.set_title("Ana Grafik Başlık")

eksen2.plot(numpyDizisi2,numpyDizisi1,"b")
eksen2.set_xlabel("X Ekseni")
eksen2.set_ylabel("Y Ekseni")
eksen2.set_title("Küçük Grafik Başlık")
```

```python
(benimFigure, benimEksen) = plt.subplots(nrows=1,ncols=2)
benimEksen.plot(numpyDizisi1,numpyDizisi2,"b")
```
