```python
# 2. Numpy Import & Veri Oluşturma

import numpy as np

# 8 çalışan x 6 ay performans verisi (50–100 arası)
performance = np.random.randint(50, 101, size=(8, 6))

performance
```

```python
# 3. Array Özellikleri

print("Shape:", performance.shape)
print("Boyut (ndim):", performance.ndim)
print("Veri tipi:", performance.dtype)
print("Toplam eleman:", performance.size)
```

```python
# 4. NumPy İndeksleme

# İlk çalışanın tüm ayları
first_employee = performance[0]
print("İlk çalışan:", first_employee)

# Son ayın tüm çalışanları
last_month = performance[:, -1]
print("Son ay:", last_month)

# İlk 3 çalışanın ilk 2 ayı
subset = performance[:3, :2]
print("Alt veri:\n", subset)
```

```python
# 5. Matrix İndeksleme

# 4. çalışanın 3. ay performansı
specific_score = performance[3, 2]
print("Belirli skor:", specific_score)

# Alt matris (çalışan 2–5, ay 1–3)
sub_matrix = performance[1:5, 1:4]
print("Sub-matrix:\n", sub_matrix)
```

```python
# 6. NumPy Operasyonlar

# Tüm skorlara +5 ekle
boosted_scores = performance + 5
print("Boosted Scores:\n", boosted_scores)

# 100 üzerinden normalize et
normalized_scores = boosted_scores / 100
print("Normalized Scores:\n", normalized_scores)

# Ay 1 ve Ay 2 karşılaştırması
comparison = performance[:, 0] > performance[:, 1]

print("Ay1 > Ay2 karşılaştırması:\n", comparison)
```

```python
# 7. NumPy Metodları
# axis=1 -> Satır boyunca (Çalışan bazlı)
employee_avg = performance.mean(axis=1)

# axis=0 -> Sütun boyunca (Ay bazlı)
month_avg = performance.mean(axis=0)

print("Çalışan ortalamaları:", employee_avg)
print("Ay ortalamaları:", month_avg)

print("En yüksek performans:", performance.max())
print("En düşük performans:", performance.min())
```

```python
# 8. Numpy Dizi Methodları
# Transpose (aylar ↔ çalışanlar)
transposed = performance.transpose()

# Tek boyuta indirgeme
flat_data = performance.flatten()

print("Transpose shape:", transposed.shape)
print("Flatten uzunluk:", len(flat_data))
```

```python
# Genel Performans Özeti
best_employee = np.argmax(employee_avg) + 1
best_month = np.argmax(month_avg) + 1
overall_avg = performance.mean()

print(f"En başarılı çalışan: Çalışan {best_employee}")
print(f"En yüksek ay ortalaması: Ay {best_month}")
print(f"Genel ortalama performans: {overall_avg:.2f}")
```
