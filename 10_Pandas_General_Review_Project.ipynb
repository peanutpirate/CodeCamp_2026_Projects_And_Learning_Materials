```python
## 2. Pandas Import & Series Oluşturma

import pandas as pd

# Aylık harcamaları temsil eden bir Series (Endeks: Ay isimleri)
months = ['Ocak', 'Şubat', 'Mart', 'Nisan', 'Mayıs', 'Haziran']
spending_values = [4200, 5100, 3800, 6200, 5500, 7100]

monthly_spending_series = pd.Series(data=spending_values, index=months)

print("Aylık Harcama Serisi:")
print(monthly_spending_series)
```

```python
## 3. Series Özellikleri

print(f"Değerler (Values): {monthly_spending_series.values}")
print(f"İndeks Yapısı: {monthly_spending_series.index}")
print(f"Veri Tipi: {monthly_spending_series.dtype}")
print(f"Ortalama Harcama: {monthly_spending_series.mean():.2f} TL")
```

```python
## 4. DataFrame Oluşturma

data = {
    'CustomerName': ['İpek', 'Can', 'Aslı', 'Mert', 'Ece', 'Emre', 'Selin', 'Arda'],
    'City': ['İstanbul', 'Ankara', 'İstanbul', 'İzmir', 'Ankara', 'İstanbul', 'İzmir', 'İstanbul'],
    'MonthlySpending': [9200, 4500, 7800, 3200, 5600, 8100, 4900, 6700],
    'MembershipType': ['Gold', 'Silver', 'Gold', 'Bronze', 'Silver', 'Gold', 'Bronze', 'Silver']
}

df_customers = pd.DataFrame(data)
print("Müşteri Veri Seti:")
print(df_customers)
```

```python
## 5. DataFrame İndeksleme

# CustomerName sütununu indeks olarak ayarlayalım
df_indexed = df_customers.set_index('CustomerName')

# Index üzerinden 'İpek' isimli müşterinin verisini seçelim
ipek_data = df_indexed.loc['İpek']

print("İndeksli DataFrame (İlk 3 Satır):")
print(df_indexed.head(3))
print("\nİpek'in Bilgileri:")
print(ipek_data)
```

```python
## 6. İndeks Değiştirme

# İndeksi sıfırlayalım (Eski indeks veri sütununa geri döner)
df_reset = df_indexed.reset_index()

# Şehir bazlı yeni bir indeks tanımlayalım (Veri olmadığını, sadece etiket olduğunu kavramak için)
df_new_index = df_reset.set_index('City')

print("Şehir İndeksli Tablo:")
print(df_new_index.head(3))
```

```python
## 7. Multi Indeks Oluşturma

# City ve MembershipType ile hiyerarşik yapı kuralım
df_multi = df_reset.set_index(['City', 'MembershipType']).sort_index()

print("Multi-Index (Şehir ve Üyelik Tipi) Yapısı:")
print(df_multi)

# Belirli bir seçim: İstanbul'daki Gold üyeler
istanbul_gold = df_multi.loc[('İstanbul', 'Gold')]
print("\nİstanbul - Gold Üyeler:")
print(istanbul_gold)
```

```python
## Genel Özet

# Hesaplamalar
total_customers = len(df_customers)
max_spending = df_customers['MonthlySpending'].max()
top_city = df_customers['City'].value_counts().idxmax()

print(f"Toplam müşteri sayısı: {total_customers}")
print(f"En yüksek aylık harcama: {max_spending} TL")
print(f"En çok müşteri olan şehir: {top_city}")
```
