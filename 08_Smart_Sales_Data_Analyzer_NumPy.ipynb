# Smart Sales Data Analyzer (NumPy v1)

## Projenin Amacı
Bu projenin amacı, NumPy kullanarak 7 günlük satış verisini analiz etmek ve
çok boyutlu array yapısını gerçek bir senaryo üzerinden öğrenmektir.

## NumPy Neden Kullanılır?
- Hızlı ve verimli sayısal işlemler için
- Çok boyutlu veri yapılarıyla çalışmak için
- Python döngülerine alternatif vektörel işlemler yapmak için

## Kullanılan NumPy Konuları
- NumPy array oluşturma
- Array özellikleri (shape, ndim, dtype)
- Indexing & slicing
- Temel NumPy metodları
- axis kullanımı
- reshape, transpose, flatten

```python
## 1. NumPy Import & Veri Oluşturma
import numpy as np

# 7 gün x 5 ürün satış verisi
sales_data = np.array([
    [120, 80, 100, 90, 110],
    [130, 85, 105, 95, 115],
    [125, 78, 98, 88, 108],
    [140, 90, 120, 100, 130],
    [150, 95, 125, 105, 135],
    [160, 100, 130, 110, 140],
    [155, 98, 128, 108, 138]
])

sales_data
```

```python
## 2. Array Özellikleri
print("Dizi Boyutu:", sales_data.shape)
print("Boyut Sayısı (ndim):", sales_data.ndim)
print("Veri Tipi:", sales_data.dtype)
print("Toplam Eleman Sayısı:", sales_data.size)
```

```python
## 3. Indexing & Slicing
# 3. günün satışları (index 2)
third_day_sales = sales_data[2]
print("3. gün satışları:", third_day_sales)

# 2. ürünün haftalık satışları (index 1)
second_product_weekly = sales_data[:, 1]
print("2. ürün haftalık satış:", second_product_weekly)

# İlk 3 gün
first_three_days = sales_data[:3]
print("İlk 3 gün:\n", first_three_days)
```

```python
## 4. Temel NumPy Methodları
total_sales = sales_data.sum()
daily_average = sales_data.mean()
max_sale = sales_data.max()
min_sale = sales_data.min()

print("Toplam haftalık satış:", total_sales)
print("Günlük ortalama satış:", daily_average)
print("En yüksek satış:", max_sale)
print("En düşük satış:", min_sale)
```

```python
## 5. Axis Kullanımı
# Gün bazlı toplam satış (axis 1 satırlar (günler) üstünde çalışır)
daily_totals = sales_data.sum(axis=1)
print("Gün bazlı toplam satış:", daily_totals)

# Ürün bazlı toplam satış (axis 2 sütunlar (ürünler) üstünde çalışır)
product_totals = sales_data.sum(axis=0)
print("Ürün bazlı toplam satış:", product_totals)
```

```python
## 6. Numpy Dizi Methodları
# reshape
reshaped_data = sales_data.reshape(5, 7)
print("Reshape (5x7):\n", reshaped_data)
# Veriyi yeniden şekillendirme (Örn: Toplam 35 veriyi 5x7 yapma - farklı bir görünüm)

# transpose
transposed_data = sales_data.T
print("Transpose:\n", transposed_data)
# Matrisi tersine çevirme (Ürünleri satır, günleri sütun yapma)

# flatten
flattened_data = sales_data.flatten()
print("Flatten:\n", flattened_data)
# Veriyi 1x35 boyutuna düzleştirme (Analiz sonrası raporlama için)
```

```python
### Genel Özet
best_selling_product = product_totals.argmax() + 1
best_day = daily_totals.argmax() + 1

print(f"Toplam haftalık satış: {total_sales}")
print(f"En çok satan ürün: Ürün {best_selling_product}")
print(f"En iyi gün: Gün {best_day}")
```
