## groupby

```python
# rowlarımızı sıralarımızı grupluyor. 
# multi indeksler gibi illa multi index olmasıa gerek yok. 
```

```python
maasSozlugu = {"Departman": ["Yazilim","Satis","Pazarlama","Finans","İnsan Kaynaklari","Hukuk"],
               "Çalışan İsimi" : ["Ahmet","Mehmet","Atil","Burak","Zeynep","Fatma"],
               "Maas" : [100,150,200,300,400,500]
              }
```

```python
maasDataFrame = pd.DataFrame(maasSozlugu)
```

```python
maasDataFrame
```

```python
grupObjesi = maasDataFrame.groupby("Departman")
```

```python
grupObjesi.count()
```

```python
grupObjesi.mean()
```

```python
grupObjesi.max
```

```python
grupObjesi.min
```

```python
grupObjesi.describe()
```
