# İş Performansı Görselleştirme Panosu (Matplotlib)

## Projenin Amacı
Bu projenin temel amacı, ham iş verilerini (satışlar, ürünler, bölgeler) karar vericiler için anlamlı görsel hikayelere dönüştürmektir.

## Neden Görselleştirme?
Veri görselleştirme, karmaşık sayısal tabloların içindeki trendleri ve anomalileri saniyeler içinde fark etmemizi sağlar. "Analiz yaptım ama anlatabiliyor muyum?" sorusuna profesyonel bir yanıttır.

## Kullanılan Teknikler
* **Line Plot:** Zaman içindeki eğilimleri görmek için.
* **Bar Plot:** Kategorik karşılaştırmalar yapmak için.
* **Subplots (İç İçe Grafikler):** Birden fazla metriği tek bir "Dashboard" görünümünde sunmak için.

```python
# 2. Veri Hazırlığı
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Aylar
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

# Aylık satışlar
monthly_sales = [120000, 135000, 160000, 150000, 170000, 190000]

# Ürün bazlı satışlar
product_sales = {
    "Product A": 320000,
    "Product B": 280000,
    "Product C": 410000,
    "Product D": 150000
}

# Bölge bazlı satışlar
region_sales = {
    "Marmara": 450000,
    "Ege": 300000,
    "İç Anadolu": 220000,
    "Akdeniz": 180000
}
```

```python
# Line Plot Zaman Serisi
plt.figure(figsize=(8, 4))

plt.plot(months, monthly_sales, marker='o', linestyle='-', color='blue')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (TL)")
plt.grid(True)

plt.show()
```

```python
# Bar Plot Ürün Karşılaştırma
plt.figure(figsize=(8, 4))

plt.bar(product_sales.keys(), product_sales.values(), color='orange')
plt.title("Product Based Sales")
plt.xlabel("Product")
plt.ylabel("Sales (TL)")

plt.show()
```

```python
# Grafik Özelleştirme
plt.figure(figsize=(8, 4))

plt.plot(
    months,
    monthly_sales,
    marker='s',
    linestyle='--',
    color='green',
    label='Monthly Sales'
)

plt.title("Customized Monthly Sales Chart")
plt.xlabel("Month")
plt.ylabel("Sales (TL)")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
# Subplots Dashboard
fig, axs = plt.subplots(2, 2, figsize=(12, 8))

# Line plot – aylık satış
axs[0, 0].plot(months, monthly_sales, marker='o')
axs[0, 0].set_title("Monthly Sales Trend")

# Bar plot – ürün bazlı
axs[0, 1].bar(product_sales.keys(), product_sales.values())
axs[0, 1].set_title("Product Performance")

# Bar plot – bölge bazlı
axs[1, 0].bar(region_sales.keys(), region_sales.values(), color='purple')
axs[1, 0].set_title("Regional Sales")

# Genel özet
axs[1, 1].text(0.1, 0.6, f"Total Sales:\n{sum(monthly_sales):,} TL", fontsize=12)
axs[1, 1].axis('off')
axs[1, 1].set_title("Executive Summary")

plt.tight_layout()
plt.show()
```

## Yönetici Yorumu

- En güçlü satış ayı: **Haziran**
- En yüksek gelir getiren ürün: **Product C**
- En güçlü bölge: **Marmara**

### Karar Destek:
- Product C için pazarlama bütçesi artırılmalı
- Marmara bölgesindeki başarılı strateji diğer bölgelere yayılmalı
- Satış trendi pozitif olduğu için Q3 hedefleri yukarı çekilebilir
